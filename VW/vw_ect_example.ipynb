```python
import vowpalwabbit

# generated from MS copilot
# Initialize VW with ECT reduction for multiclass
# Suppose we have 4 classes
vw = vowpalwabbit.Workspace("--ect 4 --cache --loss_function logistic --passes 10")

# Training examples (labels 1–4)
examples = [
    vw.parse("1 | features a b c"),
    vw.parse("2 | features d e f"),
    vw.parse("3 | features g h i"),
    vw.parse("4 | features j k l"),
]

for ex in examples:
    vw.learn(ex)

# Predict on a new example
test_ex = vw.parse("| features d k h")
prediction = vw.predict(test_ex)


print("Predicted class:", prediction)
```
