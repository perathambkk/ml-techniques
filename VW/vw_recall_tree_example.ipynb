```python
import vowpalwabbit
from vowpalwabbit import pyvw

# Configure Recalltree for 5000 classes
vw = vowpalwabbit.Workspace("--recall_tree 5000")

# Format: <label> | <namespace> <features>
# For multiclass, the label is a positive integer (1 to 5000)
training_data = "4 | features area=50 price=120000"
ex = vw.example(training_data)

# Online training pass
vw.learn(ex)

# Predict the class for a new example
test_data = "| features area=48 price=115000"
ex_test = vw.example(test_data)
prediction = vw.predict(ex_test)

print(f"Predicted Class: {prediction}")

# Clean up
vw.finish()
```
